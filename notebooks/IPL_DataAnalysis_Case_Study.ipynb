### 1.3 Identifying Missing Values

```python
# Creating a missing values summary
missing_summary = pd.DataFrame({
    'Missing Values': df.isnull().sum(),
    'Missing Percentage': (df.isnull().sum() / len(df)) * 100
})

display(missing_summary[missing_summary['Missing Values'] > 0])
```

*Observations:*

1. Missing values are present in 7 out of 20 columns.
2. The `method` column has the highest missing percentage (98.08%), with 1,074 missing values.
3. The `city` column has 51 missing values (4.66%).
4. The `result_margin` column has 19 missing values (1.74%).
5. The `winner` and `player_of_match` columns each have 5 missing values (0.46%).
6. The `target_runs` and `target_overs` columns each have 3 missing values (0.27%).
7. Missing values require column-specific investigation before deciding whether to drop, impute, or retain them.

### 1.4 Investigating Missing Values

```python
# investigating missing winner column values
missing_winners = df[df['winner'].isnull()][['date','team1','team2','result','winner']]
display(missing_winners)
```

*Observations — Missing Winner Values*

1. Five records have missing values in the `winner` column.
2. All five records have `result = 'no result'`.
3. This indicates that these matches did not produce a declared winner.
4. The missing winner values appear to be structurally missing rather than ordinary data-entry omissions.
5. We should retain these records instead of imputing an artificial winner.

```python
# investigating missing player_of_match
missing_player_of_match = df[df['player_of_match'].isnull()][['date','team1','team2','result','winner']]
display(missing_player_of_match)
```

*Observations — Missing Player of the Match*

1. All five records with missing `player_of_match` values correspond to matches with `result = 'no result'`.
2. These are the same five matches where the `winner` column is missing.
3. No winner and no Player of the Match award were recorded for these matches.
4. These missing values appear to be structurally missing rather than data-entry errors.
5. We should retain these records without artificially assigning a winner or player award.

```python
# investigating missing result_margin
missing_result_margin = df[df['result_margin'].isnull()][['date','team1','team2','result','winner','result_margin']]
display(missing_result_margin)
```

*Observations — Missing Result Margin*

1. There are 19 missing values in the `result_margin` column.
2. All 19 records belong to matches with either `tie` or `no result`.
3. 15 matches have a tied result.
4. 4 matches have no result.
5. A winning margin is not available for these match outcomes.
6. These missing values appear to be structurally missing rather than data-entry errors.
7. We should retain these records without imputing artificial winning margins.

```python
# investigating missing city
missing_city = df[df['city'].isnull()][['date','venue','city']]
display(missing_city)
```

*Observations — Missing City*

1. There are 51 missing values in the `city` column.
2. All 51 records belong to matches played at either Dubai International Cricket Stadium or Sharjah Cricket Stadium.
3. The missing city values are concentrated in matches played at these two UAE venues.
4. The venue information provides a strong clue about the missing city values.
5. These values may be recoverable using venue information, but we should verify the venue-to-city mapping before filling them.

```python
# investigating missing target_runs
missing_target_runs = df[df['target_runs'].isnull()][['date','team1','team2','result','target_runs','target_overs']]
display(missing_target_runs)
```

*Observations — Missing Target Runs and Target Overs*

1. There are 3 missing values in `target_runs`.
2. All 3 records also have missing values in `target_overs`.
3. All 3 records have `result = 'no result'`.
4. The missing target information appears to be associated with matches that produced no result.
5. These values should be retained as missing rather than filled with artificial target values.

```python
# investigating non null method values
notnull_method = df[df['method'].notnull()][['date','season','city','result','method']]
display(notnull_method)
```

*Observations — Missing Method*

1. The `method` column contains 21 non-null values.
2. All 21 non-null values are recorded as `D/L`.
3. D/L represents the Duckworth-Lewis method, used to revise targets in interrupted matches.
4. The remaining 1,074 values are missing.
5. The `method` column appears to be relevant only to matches where the D/L method was applied.
6. Missing values in this column should not automatically be treated as data-quality errors.

### Summary — Missing Value Investigation

| Column | Finding | Treatment |
|---|---|---|
| `winner` | 5 no-result matches | Retain NaN |
| `player_of_match` | Same 5 no-result matches | Retain NaN |
| `result_margin` | 19 tied/no-result matches | Retain NaN |
| `city` | 51 missing; Dubai/Sharjah venues | Verify mapping |
| `target_runs` | 3 no-result matches | Retain NaN |
| `target_overs` | Same 3 matches | Retain NaN |
| `method` | D/L recorded in 21 matches | Retain NaN |

**Conclusion:** Most missing values have a logical explanation. We will avoid unnecessary imputation and handle them based on their meaning during preprocessing.
