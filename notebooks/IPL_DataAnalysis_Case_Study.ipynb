```python
# Investigating missing city values using venue
missing_city = df[df['city'].isnull()][['venue','city']]
display(missing_city)
```

*Observation - Investigation of Missing City Values*

1. All 51 records with missing `city` values belong to two venues.
2. The venues are `Dubai International Cricket Stadium` and `Sharjah Cricket Stadium`.
3. No other venues were associated with missing city values.
4. The venue information provides a potential way to recover the missing city values.
5. The city-to-venue mapping should be verified before filling the missing values.

```python
# Verify city values associated with the two venues
venue_city_mapping = df[
    df['venue'].isin([
        'Dubai International Cricket Stadium',
        'Sharjah Cricket Stadium'
    ]) & df['city'].notnull()
][['venue', 'city']].drop_duplicates()

display(venue_city_mapping)
```

*Observation:*
1. Existing records confirm that `Dubai International Cricket Stadium` is associated with `Dubai`.
2. Existing records confirm that `Sharjah Cricket Stadium` is associated with `Sharjah`.
3. Both venue-to-city mappings are consistent in the inspected records.
4. These mappings can be used to recover the corresponding missing city values.

```python
# Assign Dubai to missing city values for Dubai International Cricket Stadium
df.loc[df['city'].isnull() & (df['venue'] == 'Dubai International Cricket Stadium'),'city'] = 'Dubai'
display(df[df['venue'] == 'Dubai International Cricket Stadium'][['venue','city']])
```

```python
# Assign Sharjah to missing city values for Sharjah Cricket Stadium
df.loc[
    df['city'].isnull() & (df['venue'] == 'Sharjah Cricket Stadium'),
    'city'
    ] = 'Sharjah'

display(df[df['venue'] == 'Sharjah Cricket Stadium'][['venue','city']])
```
